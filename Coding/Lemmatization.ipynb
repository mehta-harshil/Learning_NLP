# Lemmatization  - Text Processing

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
lemma = WordNetLemmatizer()
```

#### POS- Noun-n
#### verb-v
#### adjective-a
#### adverb-r

```python
words = ["eating","eats","eaten","writing","writes","programming","programs","testing","applied","Congratualations","going","history","goes"]

for word in words:
    print(word, " ---> ",lemma.lemmatize(word))
```
