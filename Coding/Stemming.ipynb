# Stemming


## PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
words = ["eating","eats","eaten","writing","writes","programming","programs","testing","applied","Congratualations","going","history","goes"]

for word in words:
    print(word, " ---> ",stemming.stem(word))
```

## RegxpStemmer

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

## Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
sbs = SnowballStemmer('english')
```

```python
for word in words:
    print(word," --> ",sbs.stem(word))
```

### Difference between PoterStemmer and SnowballStemmer

```python
stemming.stem('fairly'), stemming.stem('sportingly')
```

```python
sbs.stem('fairly'), sbs.stem('sportingly')
```
