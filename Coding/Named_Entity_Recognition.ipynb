# Named Entity Recognition

```python
sentence = "Founded in 1868 by Jamsetji Tata, the Tata Group is a global Indian conglomerate headquartered in Mumbai. It operates over 30 companies in 100+ countries, spanning auto, steel, IT (TCS), and consumer goods. Renowned for philanthropy, 66% of Tata Sons equity is held by philanthropic trusts."
import nltk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
import nltk 
words = nltk.word_tokenize(sentence)
tag_element = nltk.pos_tag(words)
nltk.ne_chunk(tag_element).draw()
```
