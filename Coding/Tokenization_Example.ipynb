```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
corpus = """My name is Mehta Harshil and I am persuing my bachlore’s degree in SSGEC college. I want to become Robotics Engineer.
"""
```

```python
print(corpus)
```

### Tokenization
### Paragraph - > Sentence 

```python
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize(corpus)
```

## Tokenization
## Paragraph -->  Words
## Sentence --> Words

```python
from nltk.tokenize import word_tokenize
from nltk.tokenize import wordpunct_tokenize
from nltk.tokenize import TreebankWordTokenizer
```

```python
from nltk.tokenize import word_tokenize


words1 = word_tokenize(corpus)
print(words1)


words2 = wordpunct_tokenize(corpus)
print(words2)

tokenizer = TreebankWordTokenizer()
word3 = tokenizer.tokenize(corpus)
print(word3)
```
